```python
from abc import ABC, abstractmethod
```

```python
class Runnables(ABC):

    @abstractmethod
    def invoke(input_data):
        pass
```

```python
import random

class NakliLLM(Runnables):
    def __init__(self):
        print('LLM created')
    
    def invoke(self,prompt):
        response_list = [
            
            'Delhi is the capital of India',
            'IPL is a cricket league',
            'AI stands for Artificial Intelligence'
        ]
        return {'response':random.choice(response_list)}

    def predict(self,prompt):
        response_list = [
            
            'Delhi is the capital of India',
            'IPL is a cricket league',
            'AI stands for Artificial Intelligence'
        ]
        return {'response':random.choice(response_list)}
        
```

```python
class NakliPromptTemplate(Runnables):
    def __init__(self, template,input_variables):
        self.template = template
        self.input_variables = input_variables
    
    def invoke(self,input_dict):
        return self.template.format(**input_dict)

    def format(self,input_dict):
        return self.template.format(**input_dict)
```

```python
class NakliSTRParser(Runnables):
    def __init__(self):
        pass
    def invoke(self,input_data):
        return input_data['response']
```

```python
template = NakliPromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=["topic", "length"],
)
```

```python
prompt = template.format({'length':'short','topic':'India'})
```

```python
class RunnableConnector(Runnables):
    def __init__(self,runnables_list):
        self.runnables_list = runnables_list
    
    def invoke(self,input_data):
        for runnable in self.runnables_list:
            input_data = runnable.invoke(input_data)
        return input_data
```

```python
llm = NakliLLM()

parser = NakliSTRParser()
```

```python
chain = RunnableConnector([template,llm,parser])
chain.invoke({'topic':'Black Holes','length':'short'})
```

```python
template1 = NakliPromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)
```

```python
template2 = NakliPromptTemplate(
    template='Explain the following joke {response}',
    input_variables=['response']
)
```

```python
llm = NakliLLM()
parser = NakliSTRParser()

chain1= RunnableConnector([template1,llm])
chain2= RunnableConnector([template2,llm,parser])
final_chain = RunnableConnector([chain1,chain2])

result = final_chain.invoke({"topic":"Black Holes"})
```
